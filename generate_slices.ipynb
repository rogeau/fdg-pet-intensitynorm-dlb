```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from scipy.ndimage import label

# Get original colormap
base_cmap = plt.cm.get_cmap('gist_ncar')

# Sample it
colors = base_cmap(np.linspace(0, 1, 256))

# 🔧 Modify low intensities (first ~15%)
n_low = int(0.11 * 256)

for i in range(n_low):
    t = i / (n_low - 1)  # 0 → 1

    # Take original color at boundary and scale it down
    ref_color = colors[n_low].copy()

    colors[i, :3] = t * ref_color[:3]   # scale RGB only
    colors[i, 3] = 1.0                  # keep alpha

custom_cmap = ListedColormap(colors)
```

```python
path = "shoot/HC/8007703562/20211104/wr_petsuv.nii"
img = nib.load(path)
data = img.get_fdata()
slice_axial = np.rot90(data[:, :, 50])


thr = 1.1
binary = slice_axial > thr

labeled, n = label(binary)

sizes = np.bincount(labeled.ravel())
min_size = 500

mask_clean = sizes[labeled] >= min_size
mask_clean[labeled == 0] = False

h, w = slice_axial.shape
cy, cx = h // 2, w // 2

yy, xx = np.ogrid[:h, :w]
center_mask = (yy - cy)**2 + (xx - cx)**2 <= 35**2

mask_clean = mask_clean | center_mask

slice_masked = np.ma.masked_where(~mask_clean, slice_axial)


plt.figure(figsize=(5, 5), dpi=150)

plt.imshow(
    slice_masked,
    cmap=custom_cmap,
    vmin=thr,
    vmax=11,
    interpolation="bilinear"
)

plt.axis('off')

ax = plt.gca()
ax.set_position([0, 0, 1, 1])

name = path[6:-4].replace("/", "_")

plt.savefig(
    f"shoot/summary_ips/{name}.png",
    bbox_inches='tight',
    pad_inches=0,
    transparent=True
)

plt.show()
```
